```python
import seaborn as sns
import matplotlib.pyplot as plt 
import numpy as np
# every time same data is genrated hence we seed the data
np.random.seed(42)
salary_data = np.random.exponential(scale = 20000 , size = 500) + 30000 # 

sns.set_theme(style="whitegrid")
plt.figure(figsize = (10,7))
sns.histplot(salary_data, kde = True , color = '#5e548e' , bins = 25)
plt.title("Salary Distribution Analysis", fontsize = 15 , fontweight = 'bold')
plt.xlabel("Salary ($)", fontsize = 10 , fontweight = 'bold')
plt.ylabel("No. Of Employees ", fontsize = 10 , fontweight = 'bold')
plt.show()
from scipy.stats import skew
data_skewnes = skew(salary_data)
print(f"Skewness of Salary Data is :{data_skewnes : .4f}")
```
