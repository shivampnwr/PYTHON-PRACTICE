```python
import matplotlib.pyplot as plt
import numpy as np
Students = ["Amit","Riya","John"]
Marks = [85,92,78]
```

```python
plt.figure(figsize = (8,5))
plt.bar(Students,Marks,width =0.5,color = ['#081c15','#2d6a4f','#74c69d'],label=['Amit','Riya','John'])
plt.title("Students Marks Comparison",fontweight='bold',fontsize=25,family="nimbus sans",color='#064b00')
plt.xlabel("Students",fontsize= 20 , fontweight = 'bold' , color = '#28a71e',family = 'roboto' )
plt.ylabel("Marks",fontsize= 20 , fontweight = 'bold' , color = '#28a71e',family = 'roboto' )
plt.ylim(0,100)
plt.legend()
plt.show()
```

```python
# grouped  bar graph with multiple subjects
# 1. Define the expanded dataset (multiple subjects)
students = ["AMIT", "RIYA", "JOHN"]
math_marks = [85, 92, 78]
science_marks = [88, 85, 90]
english_marks = [75, 89, 82]

# 2. Set up the x-axis positions and bar width
x_indexes = np.arange(len(students))  # Creates an array like [0, 1, 2]
width = 0.25 # The thickness of each bar
plt.figure(figsize=(8,5)) # size of figure
# 3. Plot each subject, shifting the x-position for each
plt.bar(x_indexes - width, math_marks, width=width, label='Math', color = '#FFC0CB')
plt.bar(x_indexes, science_marks, width=width, label='Science' , color = '#FF69B4')
plt.bar(x_indexes + width, english_marks, width=width, label='English', color = '#800000')

# 4. Add labels, title, and fix the x-axis ticks
plt.title("Student Performance Dashboard - Multiple Subjects",fontsize = 20 , fontweight = 'bold' )
plt.xlabel("Students" , fontsize = 15 , fontweight = 'bold')
plt.ylabel("Marks" , fontsize = 15 , fontweight = 'bold')
plt.xticks(ticks=x_indexes, labels=students) # Replaces the [0, 1, 2] with names
plt.ylim(0,120) #Stretch the Y-axis to 120 to create empty space at the top

# 5. Add a legend and display the plot
plt.legend(loc='upper left')
plt.show()
```
