```python
import os
import re
import pandas as pd
```

```python
# read job_discription
with open("job_discription.txt","r",encoding ="utf-8") as file:
    job_discription = file.read()

# skills required
required_skills = ["python","numpy","pandas","matplotlib","machine learning","sql","power bi"]

# extract inforamtion
def extract_resume_details(text):
    name_match = re.search(r"Name:\s*(.*)",text,re.IGNORECASE)
    skills_match = re.search(r"Skills:\s*(.*)",text,re.IGNORECASE)
    education_match = re.search(r"Education:\s*(.*)",text,re.IGNORECASE)
    experience_match = re.search(r"Experience:\s*(.*)",text,re.IGNORECASE)

    name = name_match.group(1).strip() if name_match else "Unknown"

    skills = skills_match.group(1).strip() if skills_match else ""


    experience = (
        experience_match.group(1).strip()
        if experience_match
        else "0 years"
    )

    education = (
        education_match.group(1).strip()
        if education_match
        else "Not specified"
    )

    return name, skills, experience, education
```

```python
# calculate match score
def calculate_score(resume_skills):
    resume_skills = resume_skills.lower()

    matched_skills = []
    missing_skills = []

    for skills in required_skills:
        if skills.lower() in resume_skills:
            matched_skills.append(skills)
        else:
            missing_skills.append(skills)

    score = len(matched_skills)/len(required_skills)*100

    return score, matched_skills, missing_skills
```

```python
# read resumes
resume_folder = "resumes"

candidates = []
for filename in os.listdir(resume_folder):
    if filename.endswith(".txt"):
        filepath = os.path.join(resume_folder,filename)

        with open(filepath, "r", encoding="utf-8") as file:
            resume_text = file.read()
            

        name, skills, experience, education = (extract_resume_details(resume_text))
        score, matched, missing = calculate_score(skills)

        candidates.append({
            "Name": name,
            "Skills": skills,
            "Experience": experience,
            "Education": education,
            "Match Score": round(score, 2),
            "Matched Skills": ", ".join(matched),
            "Missing Skills": ", ".join(missing)
        })

  # create dataframe
df = pd.DataFrame(candidates)
print(candidates)


# rank
df = df.sort_values(by ="Match Score",ascending=False)

# SHORTLISTED CANDIDATES
shortlisted = df[df["Match Score"]>60]

# export csv
shortlisted.to_csv(
    "shortlisted_candidates.csv",
    index=False
)


print("\nCandidate Ranking:")
print(df.to_string(index=False))

print("\nShortlisted Candidates:")
print(shortlisted.to_string(index=False))

print("\nShortlisted candidates saved to:")
print("shortlisted_candidates.csv")


      
```
