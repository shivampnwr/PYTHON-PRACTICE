```python
import matplotlib.pyplot as plt
```

```python
Days= ["MON","TUE","WED","THU","FRI"]
Sales = [200,250,300,280,350]
# defining the maximum and minimum values in y amd their index in x
max_sales=max(Sales)
max_day=Days[Sales.index(max_sales)]
min_sales=min(Sales)
min_day=Days[Sales.index(min_sales)]
```

```python
# creating the line graph plot with custom style lines,points,color
plt.plot(Days,Sales,marker='.',markersize=12,markerfacecolor="red",markeredgecolor="black",linestyle="dashdot",linewidth=2,label="Sales",
color='black')

# highlighting the highest and lowest values
plt.scatter(max_day,max_sales,marker='o',color= 'green',edgecolor="black",s=100, zorder=10, label='Highest Sales')
plt.scatter(min_day,min_sales,marker='o',color= 'blue',edgecolor="black",s=100, zorder=10, label='Lowest Sales')


# annotating the highest and lowest points with texts labels
plt.annotate(f'Peak: {max_sales}', xy=(max_day, max_sales), xytext=(max_day , max_sales + 15), ha='right',color="green")
plt.annotate(f'Low: {min_sales}', xy=(min_day, min_sales), xytext=(min_day , min_sales -20), ha='left',color="blue")


# customising the plot with title,xlabels,ylabels,limits
plt.title("SALES TREND VISUALIZOR",fontsize=25,family="nimbus sans",fontweight="bold",color="black")
plt.xlabel("DAYS",fontsize=20,family="Roboto Mono",fontweight="bold",color="#64748B")
plt.ylabel("SALES",fontsize=20,family="Roboto Mono",fontweight="bold",color="#64748B")
plt.ylim(150,400)

# Formatting the grid and legend
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

```
