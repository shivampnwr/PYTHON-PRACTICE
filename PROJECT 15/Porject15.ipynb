```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Student':['A','B','C','D','E','F','G','H','I','J'],
    'Math':[65,70,75,80,85,90,95,100,45,150],
    'Science':[60,68,72,78,82,88,94,96,40,140]
}

df = pd.DataFrame(data)

fig, axes = plt.subplots(2,2, figsize=(12,10))

# Line Chart
axes[0,0].plot(df['Student'], df['Math'], marker='o',color = '#d69f7e',markeredgecolor ='black',markersize =10)
axes[0,0].set_title("Math Score Trend",fontsize=20,color='#050609',fontweight='bold',family='roboto')
axes[0,0].set_xlabel("Students",fontsize=15,color='#050609',fontweight='bold',family='roboto')
axes[0,0].set_ylabel("Math Scores",fontsize=15,color='#050609',fontweight='bold',family='roboto')

# Bar Chart
axes[0,1].bar(df['Student'], df['Science'],color = '#774936',edgecolor = 'black')
axes[0,1].set_title("Science Score Comparison",fontsize=20,color='#050609',fontweight='bold',family='roboto')
axes[0,1].set_xlabel("Students",fontsize=15,color='#050609',fontweight='bold',family='roboto')
axes[0,1].set_ylabel("Science Scores",fontsize=15,color='#050609',fontweight='bold',family='roboto')

# Histogram
axes[1,0].hist(df['Math'], bins=7,color = '#f5d0c5',edgecolor = 'black')
axes[1,0].set_title("Math Score Distribution",fontsize=20,color='#050609',fontweight='bold',family='roboto')
axes[1,0].set_xlabel("Math Scores",fontsize=15,color='#050609',fontweight='bold',family='roboto')
axes[1,0].set_ylabel("Frequency",fontsize=15,color='#050609',fontweight='bold',family='roboto')

# Boxplot (Outliers) Boxplot is excellent for finding outliers. Values far from the box are suspicious. 150 will appear separately.
axes[1,1].boxplot(df['Math'])
axes[1,1].set_title("Math Outlier Detection",fontsize=20,color='#050609',fontweight='bold',family='roboto')

plt.tight_layout()

plt.show()
```
